```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("C:/Final_Capstone_Project/producer/data/fraudTrain.csv")

#=====================Data Quality Checks=======================================================================

print("DataFrame Loaded Successfully and ready for analysis.")
print("DataFrame Shape:", df.shape) # Display the shape of the DataFrame
print("DataFrame Columns:")
display(df.columns) # Display the column names
print("First 5 Rows:")
display(df.head(5)) # Display the first 5 rows of the DataFrame
print("Data Types:")
display(df.dtypes)
print("DataFrame Information:")
display(df.info())
print("Summary Statistics:")
display(df.describe())
print("Duplicate Rows:")
display(df.duplicated().sum()) # Count duplicate rows
print("Null Values:")
display(df.isnull().sum()) # Count null values in each column
print("Unique Values:")
display(df.nunique()) # Count unique values in each column
print("Value Counts for is_fraud:")    #this will give the value counts for the is_fraud column
display(df['is_fraud'].value_counts(normalize=True))
display(df['amt'].describe())  #this do a outlier analysis on the amount column
display(df.groupby('is_fraud')['amt'].describe())  #this does a outlier analysis on the amount split by the is_fraud column


#======================Top 10 breakdowns on categorical columns for fraud analysis=========================================


df['category'].value_counts().head(10) #counts how many transactions fall into each category regardless of fraud status
#df[df['is_fraud'] == 1]['category'].value_counts().head(10) #counts how many fraudulent transactions fall into each category
display(df.groupby('category')['is_fraud'].mean().sort_values(ascending=False).head(10)) #calculates the percentage of fraudulent transactions in each category and sorts them in descending order
display(df.groupby('category')['amt'].mean().sort_values(ascending=False).head(10)) #calculates the average/biggest transaction amount in each category and sorts them in descending order

merchant_counts = df['merchant'].value_counts()
valid_merchants = merchant_counts[merchant_counts >= 100].index  # only merchants with 100+ transactions
display(df[df['merchant'].isin(valid_merchants)].groupby('merchant')['is_fraud'].mean().sort_values(ascending=False).head(10))
display(df[df['merchant'].isin(valid_merchants)].groupby('merchant')['amt'].mean().sort_values(ascending=False).head(10))

df['state'].value_counts().head(10) #counts how many transactions fall into each state regardless of fraud status
display(df.groupby('state')['is_fraud'].mean().sort_values(ascending=False).head(10)) #calculates the percentage of fraudulent transactions in each state and sorts them in descending order
display(df.groupby('state')['amt'].mean().sort_values(ascending=False).head(10)) #calculates the average/biggest transaction amount in each state and sorts them in descending order

df['job'].value_counts().head(10) #counts how many transactions fall into each job regardless of fraud status
display(df.groupby('job')['is_fraud'].mean().sort_values(ascending=False).head(10)) #calculates the percentage of fraudulent transactions in each job and sorts them in descending order
display(df.groupby('job')['amt'].mean().sort_values(ascending=False).head(10)) #calculates the average/biggest transaction amount in each job and sorts them in descending order

df['gender'].value_counts().head(2) #counts how many transactions fall into each gender regardless of fraud status
display(df.groupby('gender')['is_fraud'].mean().sort_values(ascending=False).head(2)) #calculates the percentage of fraudulent transactions in each gender and sorts them in descending order
display(df.groupby('gender')['amt'].mean().sort_values(ascending=False).head(2)) #calculates the average/biggest transaction amount in each gender   


#====================columns update===================================================================

df = df.drop(columns=['Unnamed: 0'])

df = df.rename(columns={
    'trans_num': 'transaction_id',
    'trans_date_trans_time': 'event_timestamp', 
    'first': 'customer_first_name',
    'last': 'customer_last_name'
})

df.to_csv("C:/Final_Capstone_Project/producer/data/fraudTrain_clean.csv", index=False)
```
